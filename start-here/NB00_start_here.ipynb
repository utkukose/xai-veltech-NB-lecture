The next cell defines a small helper, `check`, that compares an answer with the expected value and prints feedback. Run it once; the exercises below use it.

```python
def check(name, value, expected, tol=1e-6):
    """Compares an answer with the expected value and prints immediate feedback."""
    if value is None:
        print(f"{name}: not answered yet. Replace None with your answer and run the cell again.")
        return False
    try:
        ok = abs(float(value) - float(expected)) <= tol * max(1.0, abs(float(expected)))
    except (TypeError, ValueError):
        ok = value == expected
    print(f"{name}: " + ("correct." if ok else "not yet. Compare your steps with the lecture and try again."))
    return ok
```

## Variables and types

```python
age = 21
height_m = 1.74
name = "Priya"
enrolled = True
print(type(age), type(height_m), type(name), type(enrolled))
print(f"{name} is {age} years old and {height_m:.2f} m tall.")
```

## Lists, loops and decisions

A list holds values in order. A `for` loop runs its indented block once for every element. An `if` statement runs its block only when the condition is true.

```python
temperatures = [36.4, 37.9, 36.8, 38.6, 37.1]
for t in temperatures:
    if t >= 37.5:
        print(t, "fever")
    else:
        print(t, "normal")
print("number of readings:", len(temperatures))
```

**Exercise 0.2.** Count the readings in `temperatures` that are at least 37.5 and store the count in `n_fever`.

```python
n_fever = None   # your answer
check("Exercise 0.2", n_fever, 2)
```

```python
#@title Reference solution (run this cell; open it only after trying)
n_fever = sum(1 for t in temperatures if t >= 37.5)
check("Exercise 0.2", n_fever, 2)
```

## Functions

`def` defines a reusable piece of code with parameters and a `return` value.

```python
def bmi(weight_kg, height_m):
    """Body mass index."""
    return weight_kg / height_m ** 2

print(round(bmi(68, 1.74), 1))
```

## Three libraries: NumPy, pandas and Matplotlib

`import` loads a library. NumPy computes with whole arrays at once, pandas holds tables with named columns, and Matplotlib draws plots.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

a = np.array(temperatures)
print("mean", a.mean().round(2), "max", a.max())
table = pd.DataFrame({"reading": range(1, 6), "temperature": temperatures})
print(table)
plt.plot(table["reading"], table["temperature"], "o-")
plt.axhline(37.5, color="red", linestyle="--")
plt.xlabel("reading"); plt.ylabel("temperature (°C)"); plt.show()
```

```python
#@title Reference solution (run this cell; open it only after trying)
mean_t = round(float(np.mean(temperatures)), 2)
check("Exercise 0.4", mean_t, 37.36)
```

## How Colab works

A notebook is a sequence of cells. Text cells, like this one, explain. Code cells hold Python instructions: Click into a code cell and press Shift+Enter to run it. The output appears directly below. Run the cells in order from the top. If something goes wrong, the menu Runtime, then Restart session, starts again from a clean state. The first code cell below prints a line of text and the results of three calculations: `/` divides, and `**` raises to a power.

```python
print("Hello from the start-here notebook")
print(2 + 3, 10 / 4, 2 ** 8)
```

The next cell defines a small helper, `check`, that compares an answer with the expected value and prints feedback. Run it once; the exercises below use it. Each exercise starts with `None`, the Python value for nothing, in the place of the answer, and `check` reports it as not answered yet. Reading the code of `check` is optional, since the sections below introduce most of its pieces. Two pieces appear only here. `tol=1e-6` gives the tolerance `tol` a default value, which applies when a call leaves it out. The block under `try` compares the answer as a number, and if the answer is not a number, the block under `except` compares it as it is.

```python
def check(name, value, expected, tol=1e-6):
    """Compares an answer with the expected value and prints immediate feedback."""
    if value is None:
        print(f"{name}: not answered yet. Replace None with your answer and run the cell again.")
        return False
    try:
        ok = abs(float(value) - float(expected)) <= tol * max(1.0, abs(float(expected)))
    except (TypeError, ValueError):
        ok = value == expected
    print(f"{name}: " + ("correct." if ok else "not yet. Compare your steps with the lecture and try again."))
    return ok
```

## Variables and types

A variable is a name for a value. `=` binds the value on the right to the name on the left. Numbers with a decimal point are floats, whole numbers are integers, text in quotation marks is a string, and `True` and `False` are booleans. `type` names the type of a value. The `f` before the quotation mark in the last line makes an f-string, which inserts the value of each name in braces into the text, and `:.2f` prints a number with two decimals.

```python
age = 21
height_m = 1.74
name = "Priya"
enrolled = True
print(type(age), type(height_m), type(name), type(enrolled))
print(f"{name} is {age} years old and {height_m:.2f} m tall.")
```

## Lists, loops and decisions

A list holds values in order. A `for` loop runs its indented block once for every element. An `if` statement runs its block only when the condition is true. `else` runs its block when the condition is false, and `len` counts the elements of a list.

```python
temperatures = [36.4, 37.9, 36.8, 38.6, 37.1]
for t in temperatures:
    if t >= 37.5:
        print(t, "fever")
    else:
        print(t, "normal")
print("number of readings:", len(temperatures))
```

**Exercise 0.2.** Count the readings in `temperatures` that are at least 37.5 and store the count in `n_fever`.

```python
n_fever = None   # your answer
check("Exercise 0.2", n_fever, 2)
```

```python
#@title Reference solution (run this cell; open it only after trying)
n_fever = sum(1 for t in temperatures if t >= 37.5)
check("Exercise 0.2", n_fever, 2)
```

The reference solution counts in one line. `sum(1 for t in temperatures if t >= 37.5)` adds a 1 for every reading of at least 37.5: The part inside the round brackets is a loop written as an expression, called a generator expression.

## Functions

`def` defines a reusable piece of code with parameters and a `return` value. The text in triple quotation marks describes the function, and `round(x, 1)` rounds to one decimal.

```python
def bmi(weight_kg, height_m):
    """Body mass index."""
    return weight_kg / height_m ** 2

print(round(bmi(68, 1.74), 1))
```

## Three libraries: NumPy, pandas and Matplotlib

`import` loads a library. NumPy computes with whole arrays at once, pandas holds tables with named columns, and Matplotlib draws plots. `import numpy as np` loads NumPy under the short name `np`. `np.array` turns the list into a NumPy array, and `a.mean()` and `a.max()` are methods, functions that belong to a value and are written after a dot. `range(1, 6)` counts from 1 to 5 and stops before 6. The curly braces build a dictionary, which maps keys to values, and `pd.DataFrame` turns it into a table with one column per key. `table["reading"]` selects a column by its name. In the plot, `"o-"` draws dots joined by lines, `plt.axhline` draws a horizontal line, and a semicolon puts two statements on one line.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

a = np.array(temperatures)
print("mean", a.mean().round(2), "max", a.max())
table = pd.DataFrame({"reading": range(1, 6), "temperature": temperatures})
print(table)
plt.plot(table["reading"], table["temperature"], "o-")
plt.axhline(37.5, color="red", linestyle="--")
plt.xlabel("reading"); plt.ylabel("temperature (°C)"); plt.show()
```
